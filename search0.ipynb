### ① 線形探索法（リニアサーチ）

データの先頭から順番に、目的のデータが見つかるまで一つずつ確認していく方法です。

```python
import random

# データ数の設定（デフォルトは10）
data_size = 10

# 1から100までのランダムな数字のリストを作成
data_list = random.sample(range(1, 101), data_size)
# 探したい数字をリストの中からランダムに一つ選ぶ
target = random.choice(data_list)

print(f"探すデータ: {target}")
print(f"データのリスト: {data_list}\n")

# 探索回数を数える変数
search_count = 0
# 見つかったかどうかのフラグ
found = False

# 線形探索の開始
for i in range(len(data_list)):
    search_count += 1
    current_value = data_list[i]
    print(f"{search_count}回目の確認: {i}番目のデータは {current_value} です。")

    # 目的のデータと一致するか確認
    if current_value == target:
        print(f"\nやった！ {target} を {i}番目に見つけました！")
        found = True
        break # 見つかったら繰り返しを終了

if not found:
    print(f"\n残念ながら {target} は見つかりませんでした。")

print(f"\n最終的な探索回数: {search_count}回")
```

### ② 二分探索法（バイナリサーチ）

**あらかじめデータが小さい順（または大きい順）に並んでいる必要があります。**
真ん中のデータを確認し、目的のデータがそれより大きいか小さいかで、探す範囲を半分に絞っていく方法です。データが多いときに線形探索よりも速く見つけることができます。

```python
import random

# データ数の設定（デフォルトは10）
data_size = 10

# 1から100までのランダムな数字のリストを作成し、小さい順に並び替える（ソートする）
# 二分探索ではデータが並んでいることが必須条件です！
data_list = sorted(random.sample(range(1, 101), data_size))
# 探したい数字をリストの中からランダムに一つ選ぶ
target = random.choice(data_list)

print(f"探すデータ: {target}")
print(f"小さい順に並んだリスト: {data_list}\n")

# 探索回数を数える変数
search_count = 0
# 見つかったかどうかのフラグ
found = False

# 探す範囲の最初と最後のインデックス（場所）
left = 0
right = len(data_list) - 1

# 探す範囲がなくなるまで繰り返す（leftがrightより大きくなったら終了）
while left <= right:
    search_count += 1

    # 真ん中のインデックスを計算する（// は割り算の切り捨て）
    mid = (left + right) // 2
    mid_value = data_list[mid]

    print(f"{search_count}回目の確認: 探す範囲は {left}番目から {right}番目まで。")
    print(f"  -> 真ん中の {mid}番目のデータ {mid_value} と比べます。")

    if mid_value == target:
        print(f"\nやった！ {target} を {mid}番目に見つけました！")
        found = True
        break # 見つかったら終了
    elif mid_value < target:
        # 真ん中のデータより目的のデータが大きい場合
        # 探す範囲を右半分に絞る
        print(f"  -> {target} は {mid_value} より大きいので、右半分を探します。")
        left = mid + 1
    else:
        # 真ん中のデータより目的のデータが小さい場合
        # 探す範囲を左半分に絞る
        print(f"  -> {target} は {mid_value} より小さいので、左半分を探します。")
        right = mid - 1

if not found:
    print(f"\n残念ながら {target} は見つかりませんでした。")

print(f"\n最終的な探索回数: {search_count}回")
```

二分探索法の最大の探索回数を求める計算式は、データの数を $N$$N$ とすると、数学の「対数（ログ）」を使って次のように表されます。

$$ \log_2 N $$$$ \log_2 N $$ （※より厳密に言うと、最大の回数は $\lfloor \log_2 N \rfloor + 1$$\lfloor \log_2 N \rfloor + 1$ 回となります。 $\lfloor \rfloor$$\lfloor \rfloor$ は小数点以下の切り捨てを意味します）

なぜこの式になるの？
二分探索法は、1回確認するごとに探す範囲が半分（$\frac{1}{2}$$\frac{1}{2}$）になっていきます。 1回目で $\frac{1}{2}$$\frac{1}{2}$、2回目で $\frac{1}{4}$$\frac{1}{4}$、3回目で $\frac{1}{8}$$\frac{1}{8}$ ... と減っていき、探す範囲が「残り1個」になるまで続きます。

探す回数を $k$$k$ 回とすると、最初のデータ数 $N$$N$ が $k$$k$ 回半分にされるので、 $$ N \times \left(\frac{1}{2}\right)^k = 1 $$$$ N \times \left(\frac{1}{2}\right)^k = 1 $$ となります。これを変形すると、 $$ 2^k = N $$$$ 2^k = N $$ 「2を何乗（$k$$k$乗）したら $N$$N$ になるか？」を求める計算が対数（$\log_2 N$$\log_2 N$）なので、探索回数は $\log_2 N$$\log_2 N$ で求められるというわけです。

具体例で見てみよう
$2^k$$2^k$ がデータ数 $N$$N$ を超える最初の $k$$k$ が最大探索回数になります。

データ数10の場合（今回設定した数） $2^3 = 8$、 $2^4 = 16$ なので、10を超えるには4乗が必要です。つまり最大4回で見つかります。
データ数100の場合 $2^6 = 64$、 $2^7 = 128$ なので、最大7回で見つかります。
データ数100万の場合 $2^{19} = 524,288$、 $2^{20} = 1,048,576$ なので、なんと最大20回で見つかります。
線形探索法なら100万回探さなければならないかもしれないのに、二分探索法ならたった20回で見つけられます。データが多くなればなるほど、この $\log_2 N$$\log_2 N$ の計算式が示す「圧倒的な速さ」が実感できます！
