## STEP1　分岐構造と反復構造でFizzBuzzを作る


```
# 1から30までの数字を順番に変数 i に入れて繰り返します
for i in range(1, 31):
  # もし i が 3 でも 5 でも割り切れる（＝15の倍数の）場合
  if i % 3 == 0 and i % 5 == 0:
    print("FizzBuzz")
  # そうではなく、もし i が 3 で割り切れる場合
  elif i % 3 == 0:
    print("Fizz")
  # そうではなく、もし i が 5 で割り切れる場合
  elif i % 5 == 0:
    print("Buzz")
  # 上記のどの条件にも当てはまらない場合は、数字をそのまま表示します
  else:
    print(i)
```

## STEP2　リスト内の値が偶数か奇数かを判定する

```
# 調べたい数字をまとめた「リスト」を作ります
numbers = [77, 20, 4, 31, 17]

# リストの中の数字を1つずつ取り出して、変数 x に入れて繰り返します
for x in numbers:
  # もし数字 x を 2 で割った余りが 0 の場合（＝偶数の場合）
  if x % 2 == 0:
    print(x, ":even number") # 偶数（even number）と表示します
  # そうではない場合（＝奇数の場合）
  else:
    print(x, ":odd number")  # 奇数（odd number）と表示します
```

## STEP3　５つのテストの合計と平均を求める


```
# 得点を入れるための空のリストを用意します
score = []
# 合計点を計算するための変数を用意し、最初は 0 にしておきます
sum = 0

# 5回繰り返すループを作ります（i には 0, 1, 2, 3, 4 が順番に入ります）
for i in range(5):
  # キーボードから入力された文字を整数(int)に変換して、リストの最後に追加します
  score.append(int(input('INPUT Score:')))
  # 今追加した得点（score[i]）を、これまでの合計点（sum）に足していきます
  sum = sum + score[i]

# 入力された5つの得点リストを表示します
print(score)
# 計算した合計（goukei）を表示します
print('goukei:', sum)
# 合計点をリストの要素数（今回は 5）で割って、平均（heikin）を表示します
print('heikin:', sum / len(score))
```
