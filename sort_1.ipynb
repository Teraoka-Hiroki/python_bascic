```python
# ①選択ソート (Selection Sort)
# 【仕組みの詳しい解説】
# 配列を「ソート済み部分」と「未ソート部分」に分けます。最初は全体が未ソートです。
# 未ソート部分から最小値を探し出し、未ソート部分の先頭要素と交換します。
# これを繰り返すことで、先頭から順にソート済みの要素が増えていきます。
# 【計算量】
# 最良、平均、最悪のいずれのケースでも O(N^2) となります（Nは要素数）。
# 常に未ソート部分全体をスキャンして最小値を探す必要があるためです。

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        # 未ソート部分の最小値のインデックスを見つける
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        # 最小値を未ソート部分の先頭と交換する
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        print(f"ステップ {i+1} (最小値 {arr[i]} を位置 {i} へ移動): {arr}")
    return arr

# 10個の数値
numbers1 = [29, 10, 14, 37, 13, 25, 45, 5, 18, 30]
print("=== 選択ソート ===")
print("初期状態:", numbers1)
selection_sort(numbers1)
print("完了後:", numbers1)
```

```python
# ➁バブルソート (Bubble Sort)
# 【仕組みの詳しい解説】
# 隣り合う2つの要素を比較し、順序が逆（左の方が大きい）であれば交換します。
# これを配列の先頭から末尾まで繰り返すと、1回の走査で最大値が一番右（末尾）に「浮かび上がる」ように移動します。
# これを要素数-1回繰り返すことで全体がソートされます。
# 【計算量】
# 平均および最悪ケースで O(N^2) となります。
# 最良ケース（既にソートされている場合、交換が発生したかどうかのフラグを用いる改良版の場合）は O(N) ですが、この単純な実装では O(N^2) です。

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        # 既に確定した右側の要素は比較しない
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        print(f"ステップ {i+1} (確定した最大値の位置 {n-i-1}): {arr}")
        # 交換が一度も発生しなければソート完了
        if not swapped:
            break
    return arr

# 10個の数値
numbers2 = [29, 10, 14, 37, 13, 25, 45, 5, 18, 30]
print("=== バブルソート ===")
print("初期状態:", numbers2)
bubble_sort(numbers2)
print("完了後:", numbers2)
```

```python
# ➂挿入ソート (Insertion Sort)
# 【仕組みの詳しい解説】
# トランプを配られた時に手札を順番に並べ替える方法に似ています。
# 最初の要素はソート済みとみなし、2番目の要素から順に未ソート部分から取り出します。
# 取り出した要素を、既にソート済みの部分の正しい位置に挿入していきます。
# 【計算量】
# 最悪および平均ケースで O(N^2) となります。
# ただし、データが既にほぼソートされている最良ケースでは O(N) となり、高速に動作します。

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # keyより大きい要素を一つ後ろにずらして挿入位置を確保する
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
        print(f"ステップ {i} (要素 {key} を挿入): {arr}")
    return arr

# 10個の数値
numbers3 = [29, 10, 14, 37, 13, 25, 45, 5, 18, 30]
print("=== 挿入ソート ===")
print("初期状態:", numbers3)
insertion_sort(numbers3)
print("完了後:", numbers3)
```

```python
# ④マージソート (Merge Sort)
# 【仕組みの詳しい解説】
# 「分割統治法」と呼ばれるアプローチです。
# 配列を要素数が1になるまで半分ずつに分割していきます（要素数1はソート済みとみなせるため）。
# その後、分割された配列を2つずつ、ソート順になるように「マージ（結合）」しながら元の大きさに戻していきます。
# 安定なソート（同じ値の要素の順序が保たれる）です。
# 【計算量】
# 分割に log N、マージに N かかるため、最良・平均・最悪のすべてで O(N log N) となります。
# ただし、マージ処理のために追加のメモリ空間（O(N)）を必要とするのが欠点です。

def merge_sort(arr, depth=0):
    if len(arr) > 1:
        mid = len(arr) // 2
        L = arr[:mid]
        R = arr[mid:]

        print("  " * depth + f"分割: {L} と {R}")

        merge_sort(L, depth + 1)
        merge_sort(R, depth + 1)

        i = j = k = 0
        # LとRを比較しながらマージ
        while i < len(L) and j < len(R):
            if L[i] < R[j]:
                arr[k] = L[i]
                i += 1
            else:
                arr[k] = R[j]
                j += 1
            k += 1

        while i < len(L):
            arr[k] = L[i]
            i += 1
            k += 1

        while j < len(R):
            arr[k] = R[j]
            j += 1
            k += 1

        print("  " * depth + f"マージ後: {arr}")
    return arr

# 10個の数値
numbers4 = [29, 10, 14, 37, 13, 25, 45, 5, 18, 30]
print("=== マージソート ===")
print("初期状態:", numbers4)
merge_sort(numbers4)
print("完了後:", numbers4)
```

```python
# ⑤クイックソート (Quick Sort)
# 【仕組みの詳しい解説】
# マージソートと同じく「分割統治法」です。
# 配列の中から基準値（ピボット）を1つ選び、それより「小さい要素のグループ」と「大きい要素のグループ」に分割します。
# 分割された各グループに対して、さらに再帰的にピボットを選んで分割を繰り返します。
# グループの要素数が1以下になればソート完了として結合します。
# 【計算量】
# 平均ケースで O(N log N) となり、非常に高速です。
# しかし、ピボットの選び方によっては（例えば常に最大値や最小値を選んでしまうと）、最悪ケースで O(N^2) となることがあります。

def quick_sort(arr, depth=0):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    print("  " * depth + f"ピボット[{pivot}]で分割: {left} | {middle} | {right}")

    result = quick_sort(left, depth + 1) + middle + quick_sort(right, depth + 1)

    if len(result) > 1:
         print("  " * depth + f"結合後: {result}")
    return result

# 10個の数値
numbers5 = [29, 10, 14, 37, 13, 25, 45, 5, 18, 30]
print("=== クイックソート ===")
print("初期状態:", numbers5)
# 元の配列を上書きしないよう、戻り値を受け取る
numbers5_sorted = quick_sort(numbers5)
print("完了後:", numbers5_sorted)
```
