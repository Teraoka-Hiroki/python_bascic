実際の定数で動かす



```python
# 物理シミュレーションのための初期パラメータ設定
v = 50       # 初速度 (m/s)
g = 9.8      # 重力加速度 (m/s^2)

t = 0        # 初期経過時間 (秒)
position = 0 # 初期高さ (メートル)
```

```python
# 地面に達する（高さが0未満になる）までシミュレーションを繰り返す
while position >= 0:
  t = t + 1             # 時間を1秒進める
  position = position + v # 現在の速度に応じて位置を更新
  v = v - g               # 重力加速度により速度を減算（減速・落下）
  print(t, '秒後の高さ', position)
```

現実的な物理定数で実行（0.1秒ごとに計算）

```python
# 現実的な物理定数の設定
g = 9.8      # 重力加速度 (m/s^2)
v = 50.0     # 初期速度 (m/s) ※上向きを正とする
t = 0.0      # 初期時間 (秒)
position = 0.0  # 初期高さ (m)
dt = 0.1     # シミュレーションの時間刻み幅 (秒)
```

```python
# 時間刻み dt に基づいてオイラー法でシミュレーションを実行
while position >= 0:
    # 現在の状態を出力（見やすさのため0.5秒刻み等で表示を制限する工夫も可能ですが、ここでは毎ステップ表示します）
    print(f"{t:5.1f} 秒後  高さ: {position:6.2f} m  速度: {v:5.1f} m/s")

    # 位置と速度の更新 (微小時間 dt あたりの変化量)
    position = position + v * dt
    v = v - g * dt
    t = t + dt

# 地面に達した（または突き抜けた）最終状態を出力
print(f"{t:5.1f} 秒後  高さ: {position:6.2f} m (地面に到達)")
```

物理法則（運動方程式）で計算


```python
# 現実的な物理定数の設定
g = 9.8      # 重力加速度 (m/s^2)
v = 50.0     # 初期速度 (m/s) ※上向きを正とする
t = 0.0      # 初期時間 (秒)
position = 0.0  # 初期高さ (m)
dt = 0.1     # シミュレーションの時間刻み幅 (秒)
```

```python
# 初速度を保存しておく
v0 = v

# 時間刻み dt に基づいて公式でシミュレーションを実行
while position >= 0:
    # 現在の状態を出力
    print(f"{t:5.1f} 秒後  高さ: {position:6.2f} m  速度: {v:5.1f} m/s")

    # 時間を進める
    t = t + dt

    # 物理法則の公式による位置と速度の計算 (v0を使用)
    position = v0 * t - 0.5 * g * t * t
    v = v0 - g * t

# 地面に達した（または突き抜けた）最終状態を出力
print(f"{t:5.1f} 秒後  高さ: {position:6.2f} m (地面に到達)")
```

軌跡をグラフで表示


```python
import matplotlib.pyplot as plt

# 初期値の設定
v0 = 50.0      # 初速度 (m/s)
g = 9.8        # 重力加速度 (m/s^2)
t = 0.0        # 初期時間 (s)
dt = 0.1       # 時間刻み幅 (s)
position = 0.0 # 初期高さ (m)

# グラフ描画用に、時間と位置の変化を記録するリスト
t_list = []
position_list = []

# シミュレーションを実行
while position >= 0:
    t_list.append(t)
    position_list.append(position)

    # 時間を進める
    t = t + dt

    # 物理法則の公式で位置を計算
    position = v0 * t - 0.5 * g * t * t

# 地面に到達した最後の点も追加
t_list.append(t)
position_list.append(position)

# グラフの描画設定
plt.figure(figsize=(8, 5))
plt.plot(t_list, position_list, marker='.', color='b', label='Position (m)')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.title('Time vs Position (Analytical Solution)')
plt.grid(True)
plt.legend()
plt.show()
```
