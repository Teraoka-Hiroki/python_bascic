## 例題1　じゃんけんのプログラム


```
# 1. 乱数（ランダムな数）を扱うためのライブラリを読み込みます
import random

# 2. じゃんけんの手（グー、チョキ、パー）をリストに格納します
te = ['gu', 'tyoki', 'pa']

# 3. 0から2までの整数をランダムに1つ生成し、変数aに代入します（0=gu, 1=tyoki, 2=pa）
a = random.randint(0, 2)

# 4. リスト「te」から、ランダムに選ばれたインデックス（a番目）の手を取り出して画面に表示します
print(te[a])
```

## 例題2　疑似サイコロの目の数

```
# 1. 乱数を扱うためのライブラリを読み込みます
import random

# 2. サイコロの1から6の目が出た回数を記録するためのリストを用意します（初期値はすべて0）
# インデックス0が「1の目」、インデックス5が「6の目」に対応します
dice = [0, 0, 0, 0, 0, 0]

# 3. サイコロを振る回数を1000回に設定します
i = 1000

# 4. 残り回数（i）が0より大きい間、処理を繰り返します
while i > 0:
    # 0から5までの整数をランダムに生成します（これがサイコロの目の代わりになります）
    r = random.randrange(6)
    # 出た目の位置（インデックスr）のカウントを1増やします
    dice[r] = dice[r] + 1
    # 繰り返した回数を1減らします
    i = i - 1

# 5. 最終的にそれぞれの目が出た回数のリストを画面に表示します
print(dice)
```

## 例題3　円の面積


```
# 1. 円の面積を計算する関数「circle」を定義します
# 引数（受け取る値）として半径 r を指定します
def circle(r):
    # 半径の2乗（r**2）に円周率 3.14 を掛けて、結果を返します（return）
    return r**2 * 3.14

# 2. ユーザーにキーボードから半径を入力してもらい、整数型（int）に変換して変数rに代入します
r = int(input("hankei? "))

# 3. 定義した関数「circle」を呼び出し、入力された半径 r を渡して面積を計算します
area = circle(r)

# 4. 計算された面積を画面に表示します
print(area)
```

## 追加プログラム　API

```
# 1. インターネット上の通信を行うための「requests」ライブラリを読み込みます
import requests

# 2. 検索したい郵便番号（ハイフンなしの7桁）を入力します
# 例: 東京タワーの郵便番号「1050011」など
zipcode = input("郵便番号をハイフンなしの7桁で入力してください (例: 1050011): ")

# 3. 住所データを無料提供してくれているWeb API（zipcloud）のURLを設定します
url = f"https://zipcloud.ibsnet.co.jp/api/search?zipcode={zipcode}"

# 4. requestsを使って、APIのURLにデータをリクエスト（要求）します
response = requests.get(url)

# 5. APIから返ってきた結果（JSONというデータ形式）を、Pythonの辞書形式に変換します
data = response.json()

# 6. 結果が正しく取得できたか確認して画面に表示します
if data["results"] is not None:
    # 取得したデータから都道府県、市区町村、町域名をそれぞれ取り出します
    address_data = data["results"][0]
    pref = address_data["address1"]   # 都道府県名
    city = address_data["address2"]   # 市区町村名
    town = address_data["address3"]   # 町域名

    print(f"検索結果の住所は: {pref}{city}{town} です！")
else:
    print("該当する住所が見つかりませんでした。正しい郵便番号を入力してください。")
```
